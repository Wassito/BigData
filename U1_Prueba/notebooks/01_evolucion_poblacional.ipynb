¿Qué representa la variable que analizarás? 
La variable principal, "Annual change in population" (cambio anual de población), representa el cambio neto anual de la población, calculado como la diferencia entre la población al 1 de julio de años consecutivos. Refleja de forma combinada nacimientos, defunciones y migración, y se mide en número de personas (valor entero, no un porcentaje ni una tasa de crecimiento). La serie de estimaciones cubre 1951–2023; la columna adicional "Annual population change (Projected)" contiene las proyecciones a futuro (2024–2100) basadas en el escenario medio de la ONU.

Actividad 3 — Estadística descriptiva
Propósito: aplicar medidas estadísticas básicas a una variable poblacional.

Se selecciona un año del periodo histórico (1951–2023). En este caso se elige el año 2020 y se filtra el dataset para ese año. Las medidas se calculan sobre la variable Annual change in population (cambio neto anual de población, en personas).

```python
anio = 2020
col = "Annual change in population"

hist = df[df["Year"] == anio].copy()
serie = hist[col]

hist.head()
```

```python
print("Año analizado :", anio)
```

```python
print("Media         :", serie.mean(numeric_only=True))
```

```python
print("Mediana       :", serie.median(numeric_only=True))
```

```python
print("Varianza      :", serie.var(numeric_only=True))
```

```python
print("Desv. estándar:", serie.std(numeric_only=True))
```

```python
print("Mínimo        :", serie.min(numeric_only=True))
```

```python
print("Máximo        :", serie.max(numeric_only=True))
```

Entidad con el valor mínimo y con el valor máximo

```python
ent_min = hist.loc[serie.idxmin(), "Entity"]
ent_max = hist.loc[serie.idxmax(), "Entity"]
print("Entidad con el MÍNIMO:", ent_min, "->", serie.min())
print("Entidad con el MÁXIMO:", ent_max, "->", serie.max())
```
